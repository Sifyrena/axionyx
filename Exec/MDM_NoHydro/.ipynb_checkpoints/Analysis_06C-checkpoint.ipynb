```python
from skimage.measure import marching_cubes
```

```python
tfh.plot()

tfh.tf
```

```python
import trimesh

# Example: 128x128x128 NumPy data array
data =  data_array # Replace with your actual data

# Define multiple thresholds for the "onion layers"
thresholds = 10**(np.array([3.4, 4, 4.6, 5.2, 5.8]))  # Adjust as needed

# Initialize an empty Trimesh Scene to combine layers
scene = trimesh.Scene()

# Generate and add a mesh for each threshold
for i, level in enumerate(thresholds):
    verts, faces, _, _ = marching_cubes(data, level=level)
    mesh = trimesh.Trimesh(vertices=verts, faces=faces)
    
    # Optional: Color each layer differently
    color = (PhDC.ChnArr[i,0],PhDC.ChnArr[i,1],PhDC.ChnArr[i,2], 75 - 5*i)  # RGBA
    mesh.visual.vertex_colors = color
    
    # Add to the scene
    scene.add_geometry(mesh, node_name=f"layer_{i}")

# Export the combined mesh as an OBJ or STL file
scene.export("multi_layer_mesh.obj")
```

```python
def get_N_colors(N, colormap):
    cmap = plt.get_cmap(colormap)
    return [cmap(i/N) for i in range(N)]

colormap = 'Bowls' # name of the colormap to use

cols = get_N_colors(len(TimeData), colormap)
```

```python
plt.style.use('default')

plt.rcParams['font.family'] = 'DejaVu Serif'
plt.rcParams['mathtext.fontset'] = 'dejavuserif'
plt.rcParams["text.usetex"] = False
plt.rcParams['font.size'] = 18
plt.rcParams['axes.linewidth'] = 2
plt.rcParams['lines.linewidth'] = 2
plt.rcParams['axes.facecolor'] = 'w'
```

```python
nbins = 50

middle = [0.5, 0.5, 0.5]

sphsize = (200, "kpc")

snapPath = f"../{Name}"

Palette = PhDC.CMP
```

```python
Skip = 1

fig = plt.figure(figsize = (10,20))

ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212, sharex=ax1)


Cutoff = 0 
for i, ds in enumerate(TimeData):
    
    if (i > Cutoff) and (i%Skip == 0):
        timenow = time.time()

        my_sphere = ds.sphere(middle, sphsize)

        p = yt.ProfilePlot(my_sphere,
                           "radius", ["AxDens", "density"],
                           n_bins=nbins,
                           weight_field=("boxlib", "cell_volume"),
                           accumulation=False)
        p.set_unit("radius", "kpc")
        profile = p.profiles[0]

        MaskAx = np.where(profile["AxDens"] > 0)
        MaskBa = np.where(profile["density"] > 0)

        xA = np.array(profile.x)[MaskAx]
        xB = np.array(profile.x)[MaskBa]

        yA = np.array(profile["AxDens"])[MaskAx]
        yB = np.array(profile["density"])[MaskBa]

        ax1.loglog(xA, yA, color=cols[i])
        ax2.loglog(xB, yB, color=cols[i])

        prog_bar(iteration_number=len(TimeData),
                 progress=i,
                 tinterval=time.time() - timenow,
                 status='',
                 adtl=ds)

ax1.set_ylabel("Axion Density")
ax2.set_ylabel("Baryon Density")
```

```python
ax1.set_xscale("log")
ax2.set_xscale("log")

plt.show()
```
